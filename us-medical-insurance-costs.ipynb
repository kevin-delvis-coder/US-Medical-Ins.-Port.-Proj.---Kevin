# Importing and exploring the dataset

```python
import csv
from pathlib import Path
```

```python
data_file = Path("insurance.csv")
print("dataset:", data_file.exists())
```

```python
with open(data_file, newline='') as data_file:
    reader = csv.DictReader(data_file)
    insurance_data = list(reader)
```

```python
for row in insurance_data[:5]:
    print(row)
```

```python
print(insurance_data[0]) 
```

```python
Num_of_users = len(insurance_data)
print(f"Number of users in the dataset: {Num_of_users} users.")
```

## Average age of insurance holder

```python
def avg_age(data):
    total_age = 0
    for user in data:
        total_age += int(user['age'])
    return round(total_age / len(data), 1)
```

```python
Avg_Age = avg_age(insurance_data)
print(Avg_Age)
```

## Average age of people with at least one child

```python
user_age_with_1_or_more_kids = 0
users_with_kids = 0
for user in insurance_data:
    if float(user['children']) >= 1:
        user_age_with_1_or_more_kids += float(user['age'])
        users_with_kids += 1

print(f"the average age of users with 1 or more children is {round(user_age_with_1_or_more_kids / users_with_kids)} years.")
```

# Analyzing the costs

## Average Insurance Cost

```python
def avg_cost(data):
    total_cost = 0
    for user in data:
        total_cost += float(user['charges'])
    return round(total_cost / len(data), 2)
```

```python
Avg_Insurance_Cost = avg_cost(insurance_data)
print(f"The average insurance cost is ${Avg_Insurance_Cost}")
```

## Average insurance cost by region

```python
insurance_for_region = {}
region_mentioned = {}
for user in insurance_data:
    if user['region'] in insurance_for_region:
        insurance_for_region[user['region']] += float(user['charges'])
        region_mentioned[user['region']] += 1
    else:
        insurance_for_region[user['region']] = float(user['charges'])
        region_mentioned[user['region']] = 1
for region in insurance_for_region:
    print(f"the average insurance cost for {region} is USD {round(insurance_for_region[region] / region_mentioned[region], 2)}.")
```

```python
def most_common_region(data):
    region_count = {}
    for user in data:
        region = user['region']
        if region in region_count:
            region_count[region] += 1
        else:
            region_count[region] = 1
    return region_count
common_region = most_common_region(insurance_data)
```

```python
def comm_region(data):
    region_count = 0
    most_comm_region = None
    for reg in data:
        if region_count < data[reg]:
            region_count = data[reg]
            most_comm_region = reg
    return most_comm_region, region_count
most_comm_region,region_count = comm_region(common_region)
```

```python
print(f"the most common region is {most_comm_region} with {region_count} users.")
```

# Insurance cost comparison by different factors

## Average age and insurance charges: smokers vs. nonsmokers

```python
tot_smoker = 0
smokers = 0
tot_nonsmoker = 0
nonsmokers = 0
for user in insurance_data:
    if user['smoker'] == 'yes':
        tot_smoker += float(user['charges'])
        smokers += 1
    else:
        tot_nonsmoker += float(user['charges'])
        nonsmokers += 1
avg_smoker_cost = tot_smoker / smokers
avg_nonsmoker_cost = tot_nonsmoker / nonsmokers
print(f"the average cost for smokers is USD {round(avg_smoker_cost, 2)} and for non-smokers is USD {round(avg_nonsmoker_cost, 2)}, and the difference is USD {round(avg_smoker_cost - avg_nonsmoker_cost, 2)}.")
```

```python
smoker_age = 0
smokers = 0
nonsmoker_age = 0
nonsmokers = 0
for user in insurance_data:
    if user['smoker'] == 'yes':
    
        smoker_age += float(user['age'])
        smokers += 1
    else:
        nonsmoker_age += float(user['age'])
        nonsmokers += 1
avg_smoker_age = smoker_age / smokers
avg_nonsmoker_age = nonsmoker_age / nonsmokers
print(f"the average age for smokers is  {round(avg_smoker_age)} and for non-smokers is  {round(avg_nonsmoker_age)}, and the difference is  {round(avg_smoker_age - avg_nonsmoker_age)}.")
```

## Average insurance cost of male vs female

```python
male_ins = 0
female_ins = 0
for user in insurance_data:
    if user['sex'] == 'male':
        male_ins += 1
    else:
        female_ins += 1
print(f"Number of male insured individuals: {male_ins}")
print(f"Number of female insured individuals: {female_ins}")
```

```python
insurance_by_sex = {}
for user in insurance_data:
    sex = user['sex']
    if user['sex'] in insurance_by_sex:
        insurance_by_sex[sex] += float(user['charges'])
    else:
        insurance_by_sex[sex] = float(user['charges'])
print(f"The average insurance cost for males is USD {round(insurance_by_sex['male'] / male_ins, 2)} and for females is USD {round(insurance_by_sex['female'] / female_ins, 2)}, giving a difference of USD {round((insurance_by_sex['male'] / male_ins) - (insurance_by_sex['female'] / female_ins), 2)}.")
```

## Average insurance cost of female with children vs female without vs male with children vs male without

```python
single_and_parents = {}
for user in insurance_data:
    if user['sex'] == 'female' and float(user['children']) > 0:
        if 'female with children' in single_and_parents:
            single_and_parents['female with children'] += 1
            single_and_parents['Average insurance cost female with children'] += float(user['charges'])
        else:
            single_and_parents['female with children'] = 1
            single_and_parents['Average insurance cost female with children'] = float(user['charges'])
    elif user['sex'] == 'male' and float(user['children']) > 0:
        if 'male with children' in single_and_parents:
            single_and_parents['male with children'] += 1
            single_and_parents['Average insurance cost male with children'] += float(user['charges'])
        else:
            single_and_parents['male with children'] = 1
            single_and_parents['Average insurance cost male with children'] = float(user['charges'])
    elif user['sex'] == 'female' and float(user['children']) == 0:
        if 'female without children' in single_and_parents:
            single_and_parents['female without children'] += 1
            single_and_parents['Average insurance cost female without children'] += float(user['charges'])
        else:
            single_and_parents['female without children'] = 1
            single_and_parents['Average insurance cost female without children'] = float(user['charges'])
    else:
        if 'male without children' in single_and_parents:
            single_and_parents['male without children'] += 1
            single_and_parents['Average insurance cost male without children'] += float(user['charges'])
        else:
            single_and_parents['male without children'] = 1
            single_and_parents['Average insurance cost male without children'] = float(user['charges'])

single_and_parents_avg_cost = {}
single_and_parents_avg_cost['female with children'] = round(single_and_parents['Average insurance cost female with children'] / single_and_parents['female with children'], 2)
single_and_parents_avg_cost['male with children'] = round(single_and_parents['Average insurance cost male with children'] / single_and_parents['male with children'], 2)
single_and_parents_avg_cost['female without children'] = round(single_and_parents['Average insurance cost female without children'] / single_and_parents['female without children'], 2)
single_and_parents_avg_cost['male without children'] = round(single_and_parents['Average insurance cost male without children'] / single_and_parents['male without children'], 2)
print(f"The average insurance cost for {list(single_and_parents_avg_cost.keys())[0]} is ${single_and_parents_avg_cost['female with children']}, for {list(single_and_parents_avg_cost.keys())[1]} is ${single_and_parents_avg_cost['male with children']}, for {list(single_and_parents_avg_cost.keys())[2]} is ${single_and_parents_avg_cost['female without children']}, and for {list(single_and_parents_avg_cost.keys())[3]} is ${single_and_parents_avg_cost['male without children']}.")
for user in single_and_parents_avg_cost:
    print(f"The average insurance cost for {user} is ${single_and_parents_avg_cost[user]}.")
```
